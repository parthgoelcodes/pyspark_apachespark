```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.table("workspace.default.mega_mart")
```

```python
df.limit(5).display()
```

#DROP COLUMN -- DROP()

```python
df.drop("order_date").limit(25).display()
```

# drop_duplicates 
can also writen as dropDuplicates
and also known as de drop

```python
df.dropDuplicates().limit(5).display()
```

#case - 2 USE OF SUBSET
WE CAN REMOVE DUPLICATE DATA FROM A SINGLE OR MULTIPLE ROWS


```python
df.drop_duplicates(subset = ['product_name']).display()
```

#DISTINCT ---IT WORK SAME AS DROP_DUPLICATES


```python
df.distinct().limit(5).display()
```

#DATABASE TO PRACTICE UNION

```python
data1 = [('1','kad'),
        ('2','sid')]
schema1 = 'id STRING, name STRING' 

df1 = spark.createDataFrame(data1,schema1)

data2 = [('3','rahul'),
        ('4','jas')]
schema2 = 'id STRING, name STRING' 

df2 = spark.createDataFrame(data2,schema2)
```

```python
df1.display()
```

#UNION


```python
df1.union(df2).display()
```

#THINGS THAT WILL CREATE PROBLEM

```python
data1 = [('kad','1'),
        ('sid','2')]
schema1 = 'name STRING, id STRING' 

df1 = spark.createDataFrame(data1,schema1)
```

```python
df1.union(df2).display()
```

#UNION BY MANE
IT CAN FIX THE PROBLEM 

```python
df1.unionByName(df2).display()
```

#STRING FUNCTIONS---initcap,upper,lower
INITCAP()

```python
df.select(initcap('product_name').alias('product')).limit(4).display()
```

#lower()

```python
df.select(lower('product_name')).limit(5).display()
```

#UPPER()



```python
df.select(upper('product_name')).limit(5).display()
```
